## ¿Por qué el ato anterior es semiestructurado?

Porque los datos tienen etiquetas que indican qué es cada uno, así que no es texto libre. Pero no tienen una forma fija como una tabla: hay datos dentro de otros, el número de líneas puede cambiar y no todas tienen los mismos campos.

## Tarea 8 · Clasifica
Añade una celda de texto con esta tabla y complétala:
| Dato | Tipo | Por qué |
|---|---|---|
| La tabla de tiendas de Colmena Market |Estructurado |Filas y columnas fijas todas las tiendas tienen los mismos campos. |
| Un pedido online en JSON |Semiestructurado |Tiene claves que etiquetan cada dato, pero hay datos anidados y no todos los pedidos traen los mismos campos|
| La foto de un producto |No EStructurado|la información, qué producto es, el color, etc. no viene etiquetada. |
| La reseña de un cliente |No estructurado|Es todo texto, hay que leerlo o analizarlo para sacar la opinión.|
| La grabación de una llamada a atención al cliente |No estructurado| Es audio; primero habría que transcribirlo para trabajar con él y aun asi , seria todo texto|
| El registro de accesos del servidor web |Semiestructurado|Cada línea sigue un patrón fecha, IP, URL, código, pero es texto y hay que separarlo para usarlo. |
| Una hoja de cálculo con los objetivos de ventas |Estructurado|Datos organizados en filas y columnas con un formato fijo.|
| Un correo electrónico |Semiestructurado|Tiene campos fijos remitente, destinatario, fecha, asunto, pero el cuerpo es texto libre. |
