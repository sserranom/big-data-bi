## Tarea 3 · Las cámaras de frío, con variables
Cada tienda tiene 3 cámaras de frío, y cada cámara envía su temperatura 12 veces por hora, las 24 horas del día.

Crea una variable para cada dato:
tiendas, camaras_por_tienda y lecturas_por_hora.
Calcula lecturas_al_dia a partir de ellas.
Calcula lecturas_al_anio. Los sensores no descansan: son 365 días.
Muestra los dos resultados con print y una f-string.

```python
tiendas = 120
camaras_por_tiendas = 3
lecturas_por_hora = 12
horas_por_dia = 24
dias_al_anio = 365

lecturas_al_dia = tiendas * camaras_por_tiendas * lecturas_por_hora * horas_por_dia
lecturas_al_anio = lecturas_al_dia * dias_al_anio

print(f"Lecturas del dia: {lecturas_al_dia}")
print(f"Lecturas del año: {lecturas_al_anio}")
```

## Tarea 4 · De lecturas a megas

Cada lectura ocupa unos 46 bytes. ¿Cuánto ocupan las lecturas de un día? ¿Y las de un año?

Crea la variable bytes_por_lectura.
Calcula megas_al_dia. Recuerda que 1 MB son 1.000.000 de bytes.
Calcula gigas_al_anio. Recuerda que 1 GB son 1.000 MB.
Muestra los resultados con dos decimales, usando :.2f dentro de las llaves.

```python
byte_por_lectura = 46
megas_al_dia = lecturas_al_dia * byte_por_lectura / 1000000
gigas_al_anio = megas_al_dia * dias_al_anio / 1000

print(f"Sensores: {megas_al_dia:.2f} MB al dia")
print(f"Sensores: {gigas_al_anio:.2f} GB al año")
```

## Tarea 5 · Los clics de la web y la app
La web y la app de Colmena Market reciben unas 250.000 visitas al día. En cada visita se registran unos 12 clics, y cada clic ocupa unos 220 bytes.

Calcula, con variables, cuántos clics hay al día, cuántos megas ocupan y cuántos gigas son al año.

```python
visitas_al_dia = 250000
clics_por_visitas = 12
bytes_por_clic = 220



clics_por_dia = visitas_al_dia * clics_por_visitas
megas_al_dia_por_clics = clics_por_dia * bytes_por_clic / 1000000
gigas_al_anio_por_clics = megas_al_dia_por_clics * dias_al_anio / 1000

print(f"Clics al dia: {clics_por_dia}")
print(f"Clics: {megas_al_dia_por_clics:.0f} MB al dia")
print(f"Clics: {gigas_al_anio_por_clics:.1f} GB al año")


```
